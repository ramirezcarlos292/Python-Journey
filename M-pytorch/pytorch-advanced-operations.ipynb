## 4. Efficient data processing with tf.data API

```python
from torch.utils.data import DataLoader, TensorDataset

dataset = TensorDataset(torch.arange(10).float())
dataloader = DataLoader(dataset, batch_size=2, shuffle=False)

for batch in dataloader:
    print(batch)
```

## 5. GPU Acceleration & tf.function Optimization

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

tensor = torch.tensor([1.0, 2.0, 3.0]).to(device)
print(tensor)
```

```python
@torch.jit.script
def fast_function(x: torch.Tensor):
    return x ** 2 + 3

print(fast_function(torch.tensor([2.0, 3.0])))
```
