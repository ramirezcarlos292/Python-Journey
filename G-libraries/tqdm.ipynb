# TQDM 
Creates progress bars in loops and iterables<br>
✅ Provides a dynamic progress bar for loops<br>
✅ Works with Python iterables, , and even <br>
✅ Very lightweight and easy to integrate<br>

```python
from tqdm import tqdm
import time

for i in tqdm(range(10)):
    time.sleep(0.5)
```

```python
import pandas as pd
from tqdm import tqdm

tqdm.pandas()

df = pd.DataFrame({"values": range(1000000)})
df['processed'] = df['values'].progress_apply(lambda x: x**2)
```

```python
items = list(range(100))
for item in tqdm(items, desc="Processing"):
    time.sleep(0.1)
```
