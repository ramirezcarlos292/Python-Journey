## Hashsets

```python
s = set()
print(s)

# Add items - O(1)
s.add(1)
s.add(2)
s.add(3)

print(s)
```

```python
# Look up if item in set - O(1)
if 11 in s:
    print(True)
```

```python
# Remove item from set - O(1)
s.remove(3)

print(s)
```

```python
string = "aaaaaaabbbbbbbbbbbcccccccccceeeeee"

sett = set(string)

sett
```

```python
# Loop over items in set - O(n)
for x in s:
    print(x)
```

## Hashmaps - Dictionaries

```python
d = {'greg': 1, 'steve': 2, 'rob': 3}
d
```

```python
# Add key: val in dictionary - O(1)

d['arsh'] = 4

d
```

```python
# Check for presence of key in dictionary: O(1)
if 'greg' in d:
    print(True)
```

```python
print(d['greg'])
```

```python
try:
    print(d['grg'])
except KeyError:
    print("item requested not in dictionary")
```

```python
# Loop over the key:val pairs of the dictionary: O(n)
for key, val in d.items():
    print(f"key {key}: val {val}")
    
```
