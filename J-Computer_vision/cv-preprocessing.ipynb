## Computer Vision Algorithms

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("baboon.png")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image)
plt.axis("off")
plt.show()
```

### Sharp & Blur

```python
kernel = np.array([[0, -1, 0],
                   [-1, 5,-1],
                   [0, -1, 0]])
```

```python
sharpened = cv2.filter2D(image, -1, kernel)
plt.imshow(sharpened, cmap='gray')
plt.axis('off')
plt.show()
```

```python
blurred = cv2.GaussianBlur(image, (5, 5), 0)

plt.imshow(blurred)
plt.axis('off')
plt.show()
```

### Morphological Operations

```python
kernel = np.ones((5, 5), np.uint8)

eroded = cv2.erode(image, kernel, iterations=1)
plt.imshow(eroded)
plt.axis('off')
plt.show()
```

```python
dilation = cv2.dilate(image, kernel, iterations=1)
plt.imshow(dilation, cmap='gray')
plt.axis('off')
plt.show()
```

### Edge Detection with CV2

```python
# Canny: https://docs.opencv.org/4.x/da/d22/tutorial_py_canny.html
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
edges = cv2.Canny(gray, threshold1=200, threshold2=350)

plt.imshow(edges)
plt.axis('off')
plt.show()
```

```python
# Sobel: https://docs.opencv.org/4.x/d2/d2c/tutorial_sobel_derivatives.html
sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=5) # X direction
sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=5) # Y direction
sobelxy = cv2.Sobel(gray, cv2.CV_64F, 1, 1, ksize=5) # Combined

plt.figure(figsize=(10, 5))
plt.subplot(1,3,1), plt.imshow(sobelx, cmap='gray'), plt.title('Sobel X')
plt.subplot(1,3,2), plt.imshow(sobely, cmap='gray'), plt.title('Sobel Y')
plt.subplot(1,3,3), plt.imshow(sobelxy, cmap='gray'), plt.title('Sobel XY')
plt.show()
```

```python
# Laplacian: 
laplacian = cv2.Laplacian(gray, cv2.CV_64F)

# Display result
plt.imshow(laplacian, cmap="gray")
plt.axis("off")
plt.show()
```

### Rotation

```python
height, width = image.shape[:2]

center = (width // 2, height // 2)
angle = 358
scale = 1.0
rotation_matrix = cv2.getRotationMatrix2D(center, angle, scale)

rotated_img = cv2.warpAffine(image, rotation_matrix, (width, height))

plt.imshow(rotated_img)
plt.axis('off')
plt.show()
```
