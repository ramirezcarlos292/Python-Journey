```python

from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input, decode_predictions
import numpy as np
```

```python
# 1. Load a pre-trained model (MobileNetV2 is lightweight and efficient)
model = MobileNetV2(weights='imagenet')
```

```python
# 2. Load and preprocess the image
img_path = 'baboon.png'  # Replace with the path to your image
img = image.load_img(img_path, target_size=(224, 224))  # Resize to the model's expected input size
img_array = image.img_to_array(img)
img_array_expanded_dims = np.expand_dims(img_array, axis=0)  # Add batch dimension
preprocessed_img = preprocess_input(img_array_expanded_dims)  # Preprocess according to the model's requirements
```

```python
!uname -a
```

```python
# 3. Make predictions
predictions = model.predict(preprocessed_img)
```

```python
# 4. Decode the predictions to human-readable labels
decoded_predictions = decode_predictions(predictions, top=5)  # Get the top 5 predictions
```

```python
# 5. Print the results
print('Predicted:', decoded_predictions)
```
