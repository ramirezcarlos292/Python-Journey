## Countdown

```python
def countdown(n):
    if n == 0:
        print("Blastoff!")
    else:
        print(n)
        countdown(n - 1)
        
countdown(5)
```

## Detect Palindrome

```python
def is_palindrome(word):
    if len(word) <= 1:
        return True
    elif word[0] == word[-1]:
        return is_palindrome(word[1:-1])
    else:
        return False
    
print(is_palindrome("racecar"))
print(is_palindrome("charger"))
print(is_palindrome("Anita lava la tina".lower().replace(" ", "")))
```

## Fibonacci

```python
def fibonacci(n):
    if n == 0:
        return 0
    else:
        return n + fibonacci(n - 1)

fibonacci(100)
```

## Factorial

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
    
factorial(6)
```
