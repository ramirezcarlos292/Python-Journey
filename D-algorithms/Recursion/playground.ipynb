```python
# countdown
def countdown(n):
    if n == 0:
        print("Blastoff")
    else:
        print(n)
        countdown(n - 1)
        
countdown(10)
```

```python
def palindrome(word):
    if len(word) <= 1:
        return True
    elif word[0] == word[-1]:
        return palindrome(word[1:-1])
    else:
        return False
    
print(palindrome('racecar'))
print(palindrome('charger'))
print(palindrome('Anita lava la tina'.lower().replace(" ", "")))
```

```python
def fibonacci(n):
    if n == 0:
        return 0
    else:
        return n + fibonacci(n-1)
fibonacci(100)
```

```python
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n - 1)
    
factorial(6)
```
