## Greatest Common Divisor(GCD) and Lower Common Multiplier (LCM)

```python
# GCD Euclid's algorithm
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a
print(gcd(8, 3))

# or using math.gcd
import math
print(math.gcd(8, 3))
```

```python
def lcm(a, b):
    return abs(a*b) // gcd(a, b)
print(lcm(8, 12))

# or using math.lcm()
print(math.lcm(8, 12))
```

```python
# expanding Euclid's Algorithm to three or more values
def gcd_multiple(*numbers):
    from functools import reduce
    return reduce(gcd, numbers)

print(gcd_multiple(48, 18, 30))
print(gcd_multiple(120, 45, 75, 90))
```

```python
def lcm_multiple(*numbers):
    from functools import reduce
    return reduce(lcm, numbers)

print(lcm_multiple(48, 18, 30))
print(lcm_multiple(120, 45, 75, 90))
```

# Finding Divisors of a number

- Basic divisor finder: The simplest approach is to iterate from 1 to the number and check each potential divisor.
- Optimized divisor finder: You only need to check up to the square root of the number, then you can find the corresponding divisors.
- Sieve-based approach: If you need to find divisors for multiple numbers, you can adapt a sieve of Eratosthenes approach.
- Prime factorization method: Find the prime factorization, then generate all divisors from the prime factors.

```python
# approach 1: basic divisor finder
def isThree(n):
    lst = []
    for i in range(1, n + 1):
        if n % i == 0:
            lst.append(i)
    return len(lst) == 3

print(isThree(2))
print(isThree(4))
```
