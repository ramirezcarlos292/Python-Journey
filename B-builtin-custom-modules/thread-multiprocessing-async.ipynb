```python
import time

def print_numbers():
    for i in range(1, 10):
        print(f"Number: {i}")
        time.sleep(1)
        
def print_letters():
    for letter in "ABCDE":
        print(f"Letter: {letter}")
        time.sleep(1)
```

# Threads

```python
import threading

# Create threads    
thread1 = threading.Thread(target=print_numbers)
thread2 = threading.Thread(target=print_letters)

# Start Threads
thread1.start()
thread2.start()

# Wait threads to end
thread1.join()
thread2.join()

print("Finished threading!")
```

# Multiprocessing

```python
import multiprocessing

if __name__ == "__main__":
    process1 = multiprocessing.Process(target=print_numbers)
    process2 = multiprocessing.Process(target=print_letters)
    
    process1.start()
    process2.start()
    
    process1.join()
    process2.join()
    
    print("Finished multiprocessing")
```
