```python
import tensorflow as tf
```

## 1. Creating tensors

```python
# Scalar 
scalar = tf.constant(5)

# vector
vector = tf.constant([1, 2, 3])

# Matrix
matrix = tf.constant([[1, 2], [3, 4]])


print(f"scalar: {scalar}")
print(f"vector: {vector}")
print(f"matrix: {matrix}")
```

## 2. Basic Mathematical Operations

```python
x = tf.constant([2, 3, 4])
y = tf.constant([5, 6, 7])

add = tf.add(x, y)
print(f"Addition: {add}")

sub = tf.subtract(y, x)
print(f"Substraction : {sub}")

mul = tf.multiply(x, y)
print(f"Multiply: {mul}")

div = tf.divide(y, x)
print(f"Divide: {div}")
```

## 3. Reshaping and casting

```python
tensor = tf.constant([[1, 2, 3], [4, 5, 6]])

reshaped_tensor = tf.reshape(tensor, [3, 2])

float_tensor = tf.cast(tensor, tf.float32)

print(f"Reshaped: {reshaped_tensor}")

print(f"Casted: {float_tensor}")
```

## 4. Indexing and Slicing

```python
tensor = tf.constant([10, 20, 30, 40, 50])

print(tensor[0])

print(tensor[1:4])
```
