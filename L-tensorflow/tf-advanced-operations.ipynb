## 1. Automatic Differentiation with Gradients

```python
x = tf.Variable(3.0)

with tf.GradientTape() as tape:
    y = x ** 2
grad = tape.gradient(y, x)

print(grad)
```

## 4. Efficient data processing with tf.data API
