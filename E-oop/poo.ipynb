## 1. Encapsulation
* Bundle data and methods into a single unit or class.
* Restricting direct access to object attributes using private (_ or __) variables.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
    
    def get_balance(self):
        return self.__balance
    
account = BankAccount(1000)
print(account.get_balance()) # accesible via method

print(account.__balance) # Error, Attribute is private
```

## 2. Inheritance
* Allowing a class to inherit properties and methods from another class.
* Enable code reusability

```python
class Animal:
    def speak(self):
        return "Some sound"
    
class Dog(Animal):
    def speak(self):
        return 'Bark!'
    
dog = Dog()
print(dog.speak())
```

## 3. Polymorphism
* Allowing different classes to define methods with the same name but different behavior

```python
class Bird:
    def move(self):
        return "Flies in the sky"
    
class Fish:
    def move(self):
        return "Swims in water"
    
def describe_movement(animal):
    print(animal.move())
    
describe_movement(Bird())
describe_movement(Fish())
```

## 4. Abstraction
* Hiding complex implementation details and exposing only necessary parts.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass
    
class Car(Vehicle):
    def start(self):
        return "Engine starts with a button"
    
class Bike(Vehicle):
    def start(self):
        return "Engine starts with a key"
    
print(Car().start())
print(Bike().start())
```
