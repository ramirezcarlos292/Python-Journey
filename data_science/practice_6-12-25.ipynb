```python
# Descriptive Statistics: You have the following dataset: [10, 12, 12, 15, 18, 20, 22, 22, 25, 30]. 
# Calculate the mean, median, and mode of this dataset.

data = [10, 12, 12, 15, 18, 20, 22, 22, 25, 30]
data.sort(reverse=False)
print(f"Mean: {statistics.mean(data)}")
print(f"Mode with statistics: {statistics.mode(data)}")
print(f"Mode with scipy: {scipy.stats.mode(data)}")
print(f"Median: {statistics.median(data)}")
print(f"Median: {np.median(data)}")
```

```python
# Basic Probability: A standard deck of 52 playing cards is shuffled. What is the probability of drawing a red card?

total = 52
p_take_a_red_card = 26 / total
print(f"There is a {p_take_a_red_card*100}% chance of drawing a red card")
```

```python
# Measures of Dispersion: For the dataset [2, 4, 6, 8, 10], calculate the range and the variance.
data = [2, 4, 6, 8, 10]
minn, maxx = min(data), max(data)
print(f"range:{maxx - minn}")
print(f"variance: {statistics.variance(data)}")
```

```python
# Conditional Probability (Basic): You roll a fair six-sided die. What is the probability of rolling an even number, 
# given that you rolled a number greater than 3?
print(f"Probability of rolling a number > 3 and even number is: {3/6*3/6*100}% ")
```

```python
# Expected Value (Simple): A game involves flipping a coin. If it lands on heads, you win $5. If it lands on tails, you lose $3. 
# What is the expected value of playing this game once?

5 * 0.5 + (-3 * 0.5)
```
