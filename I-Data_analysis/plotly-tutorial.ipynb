```python
import plotly.express as px
import pandas as pd
import numpy as np

# Time series example
np.random.seed(1)
df = pd.DataFrame({
    "Date": pd.date_range("2023-01-01", periods=100),
    "Sales": np.random.normal(loc=10, scale=20, size=100).cumsum()
})

fig = px.line(df, x="Date", y="Sales", title="Interactive Sales Trend")
fig.update_traces(line_color='indigo')
fig.update_layout(hovermode="x unified")
fig.show()
```

```python
import plotly.express as px

fig = px.histogram(df, x="Sales", nbins=20, color_discrete_sequence=["skyblue"])
fig.update_layout(title="Interactive Histogram of Sales")
fig.show()
```

```python
from scipy.stats import gaussian_kde

x = df["Sales"]
kde = gaussian_kde(x)
x_eval = np.linspace(x.min(), x.max(), 200)
pdf = kde(x_eval)

fig = px.line(x=x_eval, y=pdf, labels={"x": "Sales", "y": "Density"}, title="KDE Curve of Sales")
fig.update_traces(line_color="seagreen")
fig.show()
```
