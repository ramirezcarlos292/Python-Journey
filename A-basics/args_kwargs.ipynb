# *args & **kwargs

## *args

```python
def fun(*args):
    return sum(args)

print(fun(1,2,3,4))
print(fun(5,10,15))
```

```python
def fun(**kwargs):
    for k, val in kwargs.items():
        print(k, val)

fun(a=1, b=2, c=3)
```

```python
def fun(arg1, *argv):
    print("First argument: ", arg1)
    for arg in argv:
        print("Argument *argv: ", arg)
        
fun("hello", "Welcome", "to", "GFG")
```

```python
def fun(arg1, **kwargs):
    for k, val in kwargs.items():
        print("%s == %s" % (k, val))

fun("Hi", s1="Silverster", s2="Stallone", s3="Jr")
```

```python
def fun(*args, **kwargs):
    print("Positional arguments: ", args)
    print("Keyword arguments: ", kwargs)
    
fun(1, 2, 3, a=4, b=5)
```

```python
def dynamic_sum(*args):
    return sum(args)
print(dynamic_sum(1,2,3,4,5,8,10,11))
```

```python
def greet(**kwargs):
    for key, value in kwargs.items():
        print(f"{key.capitalize()}: {value}")
        
        
greet(hello="Hi There")
```

```python
"""
    - Flexible Greeting
    Create a function greet(**kwargs) where keyword arguments 
    represent different ways to greet (e.g., "hello", "hola", "bonjour"). 
    
    It should print a greeting message based on the provided keyword.
"""
def format_text(*args, **kwargs):
    formatted_text = " ".join(args)
    if kwargs.get("uppercase"):
        formatted_text = formatted_text.upper()
    if kwargs.get('lowercase'):
        formatted_text = formatted_text.lower()
    if kwargs.get('titlecase'):
        formatted_text = formatted_text.title()
        
    return formatted_text
        
        
format_text("alejandra", "carlos", "sofia", uppercase=False)
```

```python
# math operations handler
def calculate(operation, *args, **kwargs):
    
    if operation == 'add':
        calculated = 0
        for arg in args:
            calculated += arg
        
    if operation == 'multiply':
        calculated = 1
        for arg in args:
            calculated *= arg
            
    if kwargs.get('round'):
        calculated = round(calculated, 6)
    
    return calculated

print(calculate('add',9.1005, 9.445875, 5.35, round=True))
```

```python
# Dynamic Object Creator
class DynamicObject:
    def __init__(self, **kwargs):
        for key, value in kwargs.items():
            setattr(self, key, value)

    def __repr__(self):
        return f"DynamicObject({', '.join(f'{k}={v}' for k, v in self.__dict__.items())})"

# Example Usage:
obj1 = DynamicObject(name="Carlos", age=25, location="Mexico")
obj2 = DynamicObject(title="Developer", language="Python", experience=5)

print(obj1)  # Output: DynamicObject(name=Carlos, age=25, location=Mexico)
print(obj2)  # Output: DynamicObject(title=Developer, language=Python, experience=5)
```
