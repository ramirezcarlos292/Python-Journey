```python
# Import necessary libraries
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load a sample dataset (Iris)
iris = load_iris()
X = iris.data  # Features
y = iris.target  # Target variable
```

```python
# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
# Create and train the KNN model (k=5 neighbors)
model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train, y_train)
```

```python
# Make predictions
y_pred = model.predict(X_test)
```

```python
# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")
```

```python
# Example of predicting for a new data point
new_data_point = np.array([[5.1, 3.5, 1.4, 0.2]])  # Example measurements
prediction = model.predict(new_data_point)
print(f"Prediction for new data point: {iris.target_names[prediction[0]]}")
```
